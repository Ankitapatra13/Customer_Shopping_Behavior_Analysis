```python
import pandas as pd
```

```python
df = pd.read_csv("Shopping Trends And Customer Behaviour Dataset.csv")
```

```python
df.info()
```

```python
df.head()
```

```python
df = df.drop(columns="Unnamed: 0")
```

```python
df.head()
```

```python
df.describe(include="all")
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
```

```python
df = df.rename(columns={"purchase_amount_(usd)" : "purchase_amount"})
```

```python
df.columns
```

```python
# create a column age_group 
labels = ["Young Adult", "Adult", "Middle-aged", "Senior Citizen"]
df["age_group"] = pd.qcut(df["age"], q=4, labels=labels)
```

```python
df[["age","age_group"]].sample(10)
```

```python
df["frequency_of_purchases"].unique()
```

```python
# create column purchase_frequency_days
frequency_mapping = {
    "Fortnightly" : 14,
    "Weekly" : 7,
    "Annually" : 365,
    "Quarterly" : 90,
    "Bi-Weekly" : 14,
    "Monthly" : 30,
    "Every 3 Months" : 90,
}
df["purchase_frequency_days"] = df["frequency_of_purchases"].map(frequency_mapping)
```

```python
df[["frequency_of_purchases","purchase_frequency_days"]].head(10)
```

```python
df[["discount_applied","promo_code_used"]].head(10)
```

```python
# checking whether discount applied and promo code used are same ?

(df["discount_applied"] == df["promo_code_used"]).all()
```

```python
df = df.drop("promo_code_used",axis=1)
```
